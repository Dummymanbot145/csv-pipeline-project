```python
# Cell 1 - Explain the Observer Pattern simply

print("Observer Pattern has 2 parts:")
print()
print("SUBJECT  → does the work and announces results")
print("OBSERVER → listens and reacts to announcements")
print()
print("Real life example:")
print("  YouTube Channel = SUBJECT  (uploads videos)")
print("  Subscribers     = OBSERVER (get notified automatically)")
print()
print("In our CSV scenario:")
print("  SUBJECT  = Pipeline (processes files)")
print("  OBSERVER = Logger   (records results)")
print("  OBSERVER = Display  (shows results on screen)")
```

```python
# Cell 2 - The Subject: does the work and notifies observers

class Pipeline:
    def __init__(self):
        self.observers = []             # list of things watching

    def add_observer(self, observer):
        self.observers.append(observer) # add a watcher

    def notify(self, filename, result):
        for observer in self.observers: # tell every watcher what happened
            observer.update(filename, result)

    def process_file(self, filename):
        # files with "bad" in name are invalid
        result = "INVALID" if "bad" in filename else "VALID"
        print(f"Pipeline processed: {filename} → {result}")
        self.notify(filename, result)   # announce to all observers

# Test it on its own first
pipeline = Pipeline()
pipeline.process_file("sales_mar.csv")
pipeline.process_file("bad_data.csv")
```

```python
# Cell 3 - The Observers: listen and react to pipeline results

class Logger:
    name = "Logger"
    def __init__(self):
        self.log = []

    def update(self, filename, result):
        self.log.append(f"{result:<10} {filename}")
        print(f"  Logger recorded: {result} - {filename}")

class Display:
    name = "Display"

    def update(self, filename, result):
        icon = "✅" if result == "VALID" else "❌"
        print(f"  Display showed:  {icon} {filename}")

# Connect observers to pipeline and run
pipeline = Pipeline()
pipeline.add_observer(Logger())
pipeline.add_observer(Display())

print("Processing files...\n")
pipeline.process_file("sales_mar.csv")
pipeline.process_file("report_q1.csv")
pipeline.process_file("bad_data.csv")
```

```python
# Cell 4 - Everything together with a visual UI output

from IPython.display import display as show, HTML

# Subject
class Pipeline:
    def __init__(self):
        self.observers = []

    def add_observer(self, observer):
        self.observers.append(observer)

    def notify(self, filename, result):
        for observer in self.observers:
            observer.update(filename, result)

    def process_file(self, filename):
        result = "INVALID" if "bad" in filename else "VALID"
        self.notify(filename, result)

# Observer 1 - saves results to a log
class Logger:
    def __init__(self):
        self.log = []

    def update(self, filename, result):
        self.log.append((filename, result))

# Observer 2 - shows results in the UI
class HtmlDisplay:
    def __init__(self):
        self.rows = []

    def update(self, filename, result):
        color = "#4caf50" if result == "VALID" else "#ef5350"
        self.rows.append(
            f"<p style='color:{color}; font-family:monospace'>"
            f"{result:<10} {filename}</p>"
        )

    def render(self):
        show(HTML(f"""
            <div style='background:#1e1e2e; padding:20px; max-width:500px'>
                <h3 style='color:white; text-align:center'>CSV Pipeline</h3>
                <p style='color:#4caf50'>Status: Done!</p>
                <div style='background:#2e2e3e; padding:10px'>
                    {"".join(self.rows)}
                </div>
            </div>
        """))

# Run everything
pipeline = Pipeline()
logger   = Logger()
screen   = HtmlDisplay()

pipeline.add_observer(logger)   # attach logger
pipeline.add_observer(screen)   # attach display

files = ["sales_mar.csv", "report_q1.csv", "bad_data.csv"]
for f in files:
    pipeline.process_file(f)    # one call notifies both observers automatically

screen.render()                 # show the UI

print("\nLogger record:")
for filename, result in logger.log:
    print(f"  {result:<10} {filename}")
```
